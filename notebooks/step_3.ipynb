### Шаг 3. Группировка и сравнение данных

Группировка стран по регионам позволяет выявить региональные различия в уровне счастья и структуре распределений показателей, которые не видны при анализе агрегированных данных. Анализ проводится по средним значениям показателей. 

#### 3.0 Загрузка данных

```python
import pandas as pd
import seaborn as sns
import numpy as np
from matplotlib import pyplot as plt
print("Setup complete!")
```

```python
data_2015 = pd.read_csv('../data/2015.csv')
data_2015.head()
```

#### 3.1 Группировка данных по основным столбцам

Построение столбчатой диаграммы для каждого признака: уровень счастья (Happiness Score), ВВП на душу населения (Economy (GDP per Capita)), продолжительность жизни населения (Health (Life Expectancy)), восприятие коррупции населением (Trust (Government Corruption)), социальная поддержка (Family), свобода принятия решений (Freedom) и щедрость (Generosity).

#### 3.1.1 Сравнение регионов по уровню счастья

```python
regions_happiness = data_2015.groupby('Region')['Happiness Score'].mean().sort_values()
sns.barplot(x=regions_happiness.index, y=regions_happiness.values)
plt.xticks(rotation=90) 
plt.ylabel('Happiness Score')
plt.title('Средний уровень счастья по регионам')
regions_happiness
```

Средний уровень счастья в Северной Америке и Австралии и Новой Зеландии примерно одинаковый. Самый низкий уровень счастья можно наблюдать в регионе Sub-Saharan Africa. 

#### 3.1.2 Сравнение регионов по уровню ВВП на душу населения

```python
regions_economy = data_2015.groupby('Region')['Economy (GDP per Capita)'].mean().sort_values()
sns.barplot(x=regions_economy.index, y=regions_economy.values)
plt.xticks(rotation=90) 
plt.ylabel('Economy (GDP per Capita)')
plt.title('Средний уровень ВВП на душу населения по регионам')
regions_economy 
```

Лидерами по уровню ВВП на душу населения являются: Северная Америка, Западная Европа и Австралия и Новая Зеландия. Они находятся примерно на одном уровне. Самый низкий уровень демонстрирует Sub-Saharan Africa.

#### 3.1.3 Сравнение регионов по средней продолжительности жизни населения

```python
regions_health = data_2015.groupby('Region')['Health (Life Expectancy)'].mean().sort_values()
sns.barplot(x=regions_health.index, y=regions_health.values)
plt.xticks(rotation=90) 
plt.ylabel('Health (Life Expectancy)')
plt.title('Средний уровень продолжительности жизни населения по регионам')
regions_health
```

В Австралии и Новой Зеландии, Западной Европе, Северной Америке и Восточной Азии наблюдается самая высокая доля продолжительности жизни населения. Наименьшая продолжительность жизни в регионе Sub-Saharan Africa. 

#### 3.1.4 Сравнение регионов по уровню восприятия коррупции населением

```python
regions_trust = data_2015.groupby('Region')['Trust (Government Corruption)'].mean().sort_values()
sns.barplot(x=regions_trust.index, y=regions_trust.values)
plt.xticks(rotation=90) 
plt.ylabel('Trust (Government Corruption)')
plt.title('Средний уровень восприятия коррупции населением по регионам')
regions_trust
```

Регион Австралия и Новая Зеландия значительно отличается от остальных регионов - он имеет наибольшую степень доверия населения к государственным и общественным институтам. Неожиданно, но теперь самый низкий уровень показателя у Центральной и Восточной Европы. 

#### 3.1.5 Сравнение регионов по уровню социальной поддержки

```python
regions_family = data_2015.groupby('Region')['Family'].mean().sort_values()
sns.barplot(x=regions_family.index, y=regions_family.values)
plt.xticks(rotation=90) 
plt.ylabel('Family')
plt.title('Средний уровень социальной поддержки по регионам')
regions_family 
```

Социальная поддержка лучше всего у Австралии и Новой Зеландии. Хуже всего в Южной Азии.

#### 3.1.6 Сравнение регионов по уровню свободы принятия решений

```python
regions_freedom = data_2015.groupby('Region')['Freedom'].mean().sort_values()
sns.barplot(x=regions_freedom.index, y=regions_freedom.values)
plt.xticks(rotation=90) 
plt.ylabel('Freedom')
plt.title('Средний уровень свободы принятия решений по регионам')
regions_freedom
```

Лидерство в свободе принятия решений у Австралии и Новой Зеландии. Примерно на одном уровне находятся 4 региона: Центральная и Восточная Европа, Ближний Восток и Северная Африка, Sub-Saharan Africa и Южная Азия - все эти регионы имеют наименьшие показатели.  

#### 3.1.7 Сравнение регионов по уровню щедрости населения

```python
regions_generosity = data_2015.groupby('Region')['Generosity'].mean().sort_values()
sns.barplot(x=regions_generosity.index, y=regions_generosity.values)
plt.xticks(rotation=90) 
plt.ylabel('Generosity')
plt.title('Средний уровень щедрости по регионам')
regions_generosity
```

Самый высокий показатель щедрости демонстрирует Австралия и Новая Зеландия. Аутсайдером выступает Центральная и Восточная Европа. 

### Вывод

Анализ регионов показал, что самые высокие показатели чаще всего у региона Австралия и Новая Зеландия. Самые низкие измерения по основным показателям были у регионов Sub-Saharan Africa и Центральная и Восточная Европа. Однако усреднение значений, статистическое ограничение, не позволяет более объективно оценить каждый регион. То есть полученные результаты отражают общие региональные тенденции, но не индивидуальные особенности стран.

Таким образом, проведённый анализ позволил выявить ключевые факторы, статистически связанные с уровнем счастья, а также продемонстрировать различия между регионами мира.
