```python
import pandas as pd
import seaborn as sns
import numpy as np
from matplotlib import pyplot as plt
print("Setup complete!")
```

```python
data_2015 = pd.read_csv('../data/2015.csv')
data_2015.head()
```

```python
regions1 = data_2015.groupby('Region')['Happiness Score'].mean().sort_values()
sns.barplot(x=regions1.index, y=regions1.values)
plt.xticks(rotation=90) 
```
